```python
import pandas as pd
import numpy as np

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
movies = pd.read_csv('data/movies.csv')
movies
```

```python
plt.figure(figsize=(12, 8))
sns.heatmap(movies.isna(), cbar=False)
plt.title("Missing Value Heatmap")
plt.show()
```

```python
less_missing = (movies.copy())
```

```python
less_missing.loc[:, "word_count"] = (less_missing["overview"].str.split().str.len())
less_missing
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(less_missing["word_count"])

plt.title("Word Count Distribution")
plt.xlabel("Word Count")
plt.ylabel("Frequency")
plt.show()
```

```python
less_missing = less_missing[less_missing["word_count"] >= 20].copy()

less_missing
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(less_missing["word_count"])

plt.title("Word Count Distribution")
plt.xlabel("Word Count")
plt.ylabel("Frequency")
plt.show()
```

```python
less_missing["tagged_description"] = (
    less_missing[["imdb_id", "overview", "keywords"]]
    .astype(str)
    .agg(" ".join, axis=1)
)

less_missing
```

```python
less_missing.drop(["word_count"], axis=1).to_csv("data/cleaned.csv", index=False)
```
